```python
# STEP 3: Create useful date fields

df6['purchase_date'] = df6['order_purchase_timestamp'].dt.date

df6['purchase_year'] = df6['order_purchase_timestamp'].dt.year

df6['purchase_month'] = df6['order_purchase_timestamp'].dt.month

df6['purchase_month_name'] = df6['order_purchase_timestamp'].dt.month_name()

df6['purchase_year_month'] = (
    df6['order_purchase_timestamp'].dt.to_period('M').astype(str)
)

# Check the result
df6[
    [
        'order_id',
        'order_purchase_timestamp',
        'purchase_date',
        'purchase_year',
        'purchase_month',
        'purchase_month_name',
        'purchase_year_month'
    ]
].head()
```

```python
# STEP 4: Calculate delivery metrics

# Actual delivery time in days
df6['actual_delivery_days'] = (
    df6['order_delivered_customer_date'] -
    df6['order_purchase_timestamp']
).dt.total_seconds() / (24 * 60 * 60)

# Estimated delivery time in days
df6['estimated_delivery_days'] = (
    df6['order_estimated_delivery_date'] -
    df6['order_purchase_timestamp']
).dt.total_seconds() / (24 * 60 * 60)

# Difference between actual and estimated delivery
df6['delivery_variance_days'] = (
    df6['order_delivered_customer_date'] -
    df6['order_estimated_delivery_date']
).dt.total_seconds() / (24 * 60 * 60)

# Delivery status
df6['delivery_status'] = None

df6.loc[
    df6['order_delivered_customer_date'].notna() &
    (df6['delivery_variance_days'] <= 0),
    'delivery_status'
] = 'On Time'

df6.loc[
    df6['order_delivered_customer_date'].notna() &
    (df6['delivery_variance_days'] > 0),
    'delivery_status'
] = 'Late'
```

```python
# ============================================
# PHASE 3.12C — CUSTOMER RECENCY
# ============================================

reference_date = (
    fact_order['order_purchase_timestamp'].max()
    + pd.Timedelta(days=1)
)

customer_analytics['recency'] = (
    reference_date
    - customer_analytics['last_purchase_date']
).dt.days

print("Reference date:", reference_date)

print("\nRecency statistics:")
display(customer_analytics['recency'].describe())

print("\nNegative recency:",
      (customer_analytics['recency'] < 0).sum())

print("\nMissing recency:",
      customer_analytics['recency'].isna().sum())

display(
    customer_analytics[
        [
            'customer_unique_id',
            'first_purchase_date',
            'last_purchase_date',
            'order_count',
            'monetary_value',
            'item_count',
            'recency'
        ]
    ].head(10)
)
```

```python
# ============================================
# STEP 3.12D — RFM SCORING
# ============================================

# Rank first so qcut creates stable 5 groups even when many values are tied
customer_analytics['r_score'] = pd.qcut(
    customer_analytics['recency'].rank(method='first'),
    q=5,
    labels=[5, 4, 3, 2, 1]
).astype(int)

customer_analytics['f_score'] = pd.qcut(
    customer_analytics['order_count'].rank(method='first'),
    q=5,
    labels=[1, 2, 3, 4, 5]
).astype(int)

customer_analytics['m_score'] = pd.qcut(
    customer_analytics['monetary_value'].rank(method='first'),
    q=5,
    labels=[1, 2, 3, 4, 5]
).astype(int)

# Overall RFM score: 3 to 15
customer_analytics['rfm_score'] = (
    customer_analytics['r_score'] +
    customer_analytics['f_score'] +
    customer_analytics['m_score']
)

print("========== RFM SCORING ==========")
print("Customers:", len(customer_analytics))

print("\nR Score distribution:")
print(customer_analytics['r_score'].value_counts().sort_index())

print("\nF Score distribution:")
print(customer_analytics['f_score'].value_counts().sort_index())

print("\nM Score distribution:")
print(customer_analytics['m_score'].value_counts().sort_index())

print("\nRFM Score range:")
print(
    "Min:", customer_analytics['rfm_score'].min(),
    "| Max:", customer_analytics['rfm_score'].max()
)

print("\nRFM Score distribution:")
print(customer_analytics['rfm_score'].value_counts().sort_index())
```

```python
# ============================================
# STEP 3.12E — CUSTOMER SEGMENTATION
# ============================================

def assign_customer_segment(row):
    r = row['r_score']
    f = row['f_score']
    m = row['m_score']

    # Strong recent, frequent and high-value customers
    if r >= 4 and f >= 4 and m >= 4:
        return 'Champions'

    # Frequent and reasonably recent/high-value customers
    elif r >= 3 and f >= 4 and m >= 3:
        return 'Loyal Customers'

    # Recent customers showing potential to become loyal
    elif r >= 4 and f >= 2 and f <= 3 and m >= 2:
        return 'Potential Loyalists'

    # Very recent customers with limited purchase history
    elif r == 5 and f == 1:
        return 'New Customers'

    # Previously engaged/high-value customers who are becoming inactive
    elif r <= 2 and f >= 3 and m >= 3:
        return 'At Risk'

    # Customers needing attention but not yet strongly at risk
    elif r <= 3 and f <= 2 and m >= 2:
        return 'Need Attention'

    # Low recency, low frequency and low monetary value
    elif r <= 2 and f <= 2 and m <= 2:
        return 'Hibernating'

    # Remaining customers
    else:
        return 'Other'


customer_analytics['customer_segment'] = (
    customer_analytics.apply(assign_customer_segment, axis=1)
)

print("========== CUSTOMER SEGMENTS ==========")

segment_counts = (
    customer_analytics['customer_segment']
    .value_counts()
    .rename_axis('customer_segment')
    .reset_index(name='customers')
)

segment_counts['percentage'] = (
    segment_counts['customers'] /
    len(customer_analytics) * 100
).round(2)

print(segment_counts.to_string(index=False))
```

```python
# ============================================
# OTHER — RFM COMBINATION ANALYSIS
# ============================================

other_combinations = (
    other_customers
    .groupby(['r_score', 'f_score', 'm_score'])
    .size()
    .reset_index(name='customers')
    .sort_values('customers', ascending=False)
)

other_combinations['percentage_of_other'] = (
    other_combinations['customers'] /
    len(other_customers) * 100
).round(2)

print("========== OTHER RFM COMBINATIONS ==========")

print(
    other_combinations.head(20).to_string(index=False)
)
```

```python
# ============================================
# STEP 3.12G — FINAL CUSTOMER SEGMENTATION
# ============================================

def assign_customer_segment(row):
    r = row['r_score']
    f = row['f_score']
    m = row['m_score']

    # 1. Champions
    if r >= 4 and f >= 4 and m >= 4:
        return 'Champions'

    # 2. Loyal Customers
    elif r >= 3 and f >= 4 and m >= 3:
        return 'Loyal Customers'

    # 3. Potential Loyalists
    elif r >= 4 and f >= 2 and m >= 2:
        return 'Potential Loyalists'

    # 4. New Customers
    elif r >= 4 and f == 1:
        return 'New Customers'

    # 5. At Risk
    elif r <= 2 and f >= 3 and m >= 3:
        return 'At Risk'

    # 6. Hibernating
    elif r <= 2 and f <= 2 and m <= 2:
        return 'Hibernating'

    # 7. Need Attention
    elif r <= 3 and f <= 2:
        return 'Need Attention'

    # 8. At-risk / lower-value customers
    else:
        return 'Needs Re-engagement'


customer_analytics['customer_segment'] = (
    customer_analytics.apply(assign_customer_segment, axis=1)
)

print("========== FINAL CUSTOMER SEGMENTS ==========")

segment_counts = (
    customer_analytics['customer_segment']
    .value_counts()
    .rename_axis('customer_segment')
    .reset_index(name='customers')
)

segment_counts['percentage'] = (
    segment_counts['customers'] /
    len(customer_analytics) * 100
).round(2)

print(segment_counts.to_string(index=False))
```

```python
# ============================================
# STEP 3.13A — SELLER ANALYTICS BASE
# ============================================

seller_analytics = (
    fact_order_item
    .groupby('seller_id')
    .agg(
        orders=('order_id', 'nunique'),
        items_sold=('order_item_id', 'count'),
        product_revenue=('price', 'sum'),
        freight_value=('freight_value', 'sum')
    )
    .reset_index()
)

# Total order value
seller_analytics['order_value'] = (
    seller_analytics['product_revenue'] +
    seller_analytics['freight_value']
)

# Average order value
seller_analytics['aov'] = (
    seller_analytics['product_revenue'] /
    seller_analytics['orders']
)

print("========== SELLER ANALYTICS BASE ==========")

print("Shape:", seller_analytics.shape)

print(
    "Unique sellers:",
    seller_analytics['seller_id'].nunique()
)

print(
    "Duplicate sellers:",
    seller_analytics['seller_id'].duplicated().sum()
)

print("\nMissing values:")
print(seller_analytics.isna().sum())

print("\nSample:")
print(seller_analytics.head())
```

```python
# ============================================
# STEP 3.13B — SELLER DELIVERY PERFORMANCE
# ============================================

# Get seller + order relationship
seller_orders = (
    fact_order_item[['seller_id', 'order_id']]
    .drop_duplicates()
)

# Attach order-level delivery information
seller_orders = seller_orders.merge(
    fact_order[
        [
            'order_id',
            'order_status',
            'actual_delivery_days',
            'delivery_status'
        ]
    ],
    on='order_id',
    how='left'
)

# Keep only delivered orders with a valid delivery duration
valid_delivery = seller_orders[
    (seller_orders['order_status'] == 'delivered') &
    (seller_orders['actual_delivery_days'].notna())
].copy()

# Aggregate delivery metrics by seller
seller_delivery = (
    valid_delivery
    .groupby('seller_id')
    .agg(
        delivered_orders=('order_id', 'nunique'),
        avg_delivery_days=('actual_delivery_days', 'mean'),
        late_orders=(
            'delivery_status',
            lambda x: (x == 'Late').sum()
        )
    )
    .reset_index()
)

# Late delivery rate
seller_delivery['late_delivery_rate'] = (
    seller_delivery['late_orders'] /
    seller_delivery['delivered_orders'] * 100
)

# Merge into seller analytics
seller_analytics = seller_analytics.merge(
    seller_delivery,
    on='seller_id',
    how='left'
)

print("========== SELLER DELIVERY PERFORMANCE ==========")

print("Shape:", seller_analytics.shape)

print("\nMissing values:")
print(seller_analytics[
    [
        'delivered_orders',
        'avg_delivery_days',
        'late_orders',
        'late_delivery_rate'
    ]
].isna().sum())

print("\nDelivery metrics:")
print(
    seller_analytics[
        [
            'delivered_orders',
            'avg_delivery_days',
            'late_orders',
            'late_delivery_rate'
        ]
    ].describe()
)

print("\nSample:")
print(seller_analytics.head())
```

```python
# ============================================
# STEP 3.13C — SELLER REVIEW PERFORMANCE
# ============================================

# Unique seller-order relationship
seller_review_orders = (
    fact_order_item[['seller_id', 'order_id']]
    .drop_duplicates()
)

# Attach order-level review metrics
seller_review_orders = seller_review_orders.merge(
    fact_order[
        [
            'order_id',
            'review_count',
            'avg_review_score'
        ]
    ],
    on='order_id',
    how='left'
)

# Keep orders that actually have review information
valid_reviews = seller_review_orders[
    seller_review_orders['avg_review_score'].notna()
].copy()

# Aggregate reviews by seller
seller_reviews = (
    valid_reviews
    .groupby('seller_id')
    .agg(
        reviewed_orders=('order_id', 'nunique'),
        avg_review_score=('avg_review_score', 'mean')
    )
    .reset_index()
)

# Merge review metrics into seller analytics
seller_analytics = seller_analytics.merge(
    seller_reviews,
    on='seller_id',
    how='left'
)

print("========== SELLER REVIEW PERFORMANCE ==========")

print("Shape:", seller_analytics.shape)

print("\nMissing review metrics:")
print(
    seller_analytics[
        [
            'reviewed_orders',
            'avg_review_score'
        ]
    ].isna().sum()
)

print("\nReview metrics:")
print(
    seller_analytics[
        [
            'reviewed_orders',
            'avg_review_score'
        ]
    ].describe()
)

print("\nReview score range:")
print(
    "Min:",
    seller_analytics['avg_review_score'].min(),
    "| Max:",
    seller_analytics['avg_review_score'].max()
)

print("\nSample:")
print(
    seller_analytics[
        [
            'seller_id',
            'orders',
            'delivered_orders',
            'late_delivery_rate',
            'reviewed_orders',
            'avg_review_score'
        ]
    ].head()
)
```

```python
# ============================================
# STEP 7.4 — SELLER CUSTOMER SATISFACTION
# ============================================

# Create seller-level review metrics from the
# unique seller-order relationship

seller_order_reviews = (
    fact_order_item[
        ['seller_id', 'order_id']
    ]
    .drop_duplicates()
    .merge(
        fact_order[
            ['order_id', 'avg_review_score']
        ],
        on='order_id',
        how='left'
    )
)

seller_review_metrics = (
    seller_order_reviews[
        seller_order_reviews['avg_review_score'].notna()
    ]
    .groupby('seller_id')
    .agg(
        review_count=('order_id', 'nunique'),
        avg_review_score=('avg_review_score', 'mean')
    )
    .reset_index()
)

# Start from existing seller analytics
seller_review_analysis = (
    seller_analytics[
        [
            'seller_id',
            'orders',
            'items_sold',
            'product_revenue',
            'aov',
            'delivered_orders',
            'avg_delivery_days',
            'late_orders',
            'late_delivery_rate'
        ]
    ]
    .copy()
)

# Add review metrics
seller_review_analysis = seller_review_analysis.merge(
    seller_review_metrics,
    on='seller_id',
    how='left',
    validate='one_to_one'
)

# Revenue contribution
seller_review_analysis['revenue_share_pct'] = (
    seller_review_analysis['product_revenue']
    / seller_review_analysis['product_revenue'].sum()
    * 100
)

# Use 50+ reviewed orders to avoid unstable averages
qualified_review_sellers = seller_review_analysis[
    seller_review_analysis['review_count'] >= 50
].copy()

qualified_review_sellers = (
    qualified_review_sellers
    .sort_values('avg_review_score', ascending=True)
    .reset_index(drop=True)
)

print("========== SELLER CUSTOMER SATISFACTION ==========")

print(
    qualified_review_sellers.head(20).to_string(index=False)
)

print("\n========== QUALIFIED SELLER SUMMARY ==========")

print(
    "Total sellers:",
    seller_review_analysis['seller_id'].nunique()
)

print(
    "Sellers with 50+ reviews:",
    qualified_review_sellers['seller_id'].nunique()
)

print(
    "Average seller review score:",
    qualified_review_sellers['avg_review_score'].mean()
)

print(
    "Median seller review score:",
    qualified_review_sellers['avg_review_score'].median()
)

print(
    "Minimum seller review score:",
    qualified_review_sellers['avg_review_score'].min()
)

print(
    "Maximum seller review score:",
    qualified_review_sellers['avg_review_score'].max()
)

print("\n========== VALIDATION ==========")

print(
    "Expected sellers:",
    seller_analytics['seller_id'].nunique()
)

print(
    "Duplicate seller IDs:",
    seller_review_analysis['seller_id'].duplicated().sum()
)

print(
    "Missing seller IDs:",
    seller_review_analysis['seller_id'].isna().sum()
)

print(
    "Sellers with no review data:",
    seller_review_analysis['avg_review_score'].isna().sum()
)

print(
    "Review metrics duplicate seller IDs:",
    seller_review_metrics['seller_id'].duplicated().sum()
)
```

```python
# ============================================
# STEP 7.5 — SELLER OPERATIONAL RISK
# ============================================

risk_base = seller_review_analysis.copy()

# Use sellers with sufficient operational evidence
risk_base = risk_base[
    (risk_base['delivered_orders'] >= 50) &
    (risk_base['review_count'] >= 50)
].copy()

# Data-driven thresholds
revenue_median = risk_base['product_revenue'].median()
late_rate_median = risk_base['late_delivery_rate'].median()
review_median = risk_base['avg_review_score'].median()

print("========== DATA-DRIVEN THRESHOLDS ==========")

print(
    f"Revenue median: ₹{revenue_median:,.2f}"
)

print(
    f"Late-delivery-rate median: {late_rate_median:.2f}%"
)

print(
    f"Review-score median: {review_median:.2f}"
)

# Risk flags
risk_base['high_revenue'] = (
    risk_base['product_revenue'] >= revenue_median
)

risk_base['high_late_rate'] = (
    risk_base['late_delivery_rate'] >= late_rate_median
)

risk_base['low_review_score'] = (
    risk_base['avg_review_score'] <= review_median
)

# Operational risk score
risk_base['risk_score'] = (
    risk_base['high_late_rate'].astype(int)
    +
    risk_base['low_review_score'].astype(int)
)

# Business classification
risk_base['seller_risk_category'] = np.select(
    [
        risk_base['high_revenue'] &
        (risk_base['risk_score'] >= 2),

        risk_base['high_revenue'] &
        (risk_base['risk_score'] == 1),

        (~risk_base['high_revenue']) &
        (risk_base['risk_score'] >= 2),

        (~risk_base['high_revenue']) &
        (risk_base['risk_score'] == 1)
    ],
    [
        'High Revenue / Operational Risk',
        'High Revenue / Monitor',
        'Lower Revenue / Operational Risk',
        'Lower Revenue / Monitor'
    ],
    default='High Revenue / Reliable'
)

print("\n========== SELLER RISK DISTRIBUTION ==========")

print(
    risk_base['seller_risk_category']
    .value_counts()
)

print("\n========== HIGH REVENUE / OPERATIONAL RISK ==========")

high_risk_sellers = (
    risk_base[
        risk_base['seller_risk_category']
        == 'High Revenue / Operational Risk'
    ]
    .sort_values(
        ['product_revenue', 'late_delivery_rate'],
        ascending=[False, False]
    )
)

print(
    high_risk_sellers[
        [
            'seller_id',
            'orders',
            'delivered_orders',
            'product_revenue',
            'aov',
            'late_delivery_rate',
            'avg_delivery_days',
            'review_count',
            'avg_review_score',
            'revenue_share_pct',
            'risk_score',
            'seller_risk_category'
        ]
    ]
    .head(20)
    .to_string(index=False)
)

print("\n========== VALIDATION ==========")

print(
    "Risk analysis sellers:",
    risk_base['seller_id'].nunique()
)

print(
    "Duplicate seller IDs:",
    risk_base['seller_id'].duplicated().sum()
)

print(
    "Missing seller IDs:",
    risk_base['seller_id'].isna().sum()
)

print(
    "Risk score range:",
    risk_base['risk_score'].min(),
    "to",
    risk_base['risk_score'].max()
)

print(
    "Missing risk categories:",
    risk_base['seller_risk_category'].isna().sum()
)
```

```python
# ============================================
# STEP 7.6 — FINAL SELLER SEGMENTATION
# ============================================

seller_segmentation = risk_base.copy()

# Convert the risk category into a final
# business-friendly seller segment.

seller_segmentation['seller_segment'] = (
    seller_segmentation['seller_risk_category']
)

print("========== FINAL SELLER SEGMENTS ==========")

segment_summary = (
    seller_segmentation
    .groupby('seller_segment')
    .agg(
        sellers=('seller_id', 'nunique'),
        total_revenue=('product_revenue', 'sum'),
        avg_revenue=('product_revenue', 'mean'),
        median_revenue=('product_revenue', 'median'),
        avg_late_rate=('late_delivery_rate', 'mean'),
        avg_delivery_days=('avg_delivery_days', 'mean'),
        avg_review_score=('avg_review_score', 'mean'),
        total_orders=('orders', 'sum'),
        total_delivered_orders=('delivered_orders', 'sum')
    )
    .reset_index()
    .sort_values('total_revenue', ascending=False)
    .reset_index(drop=True)
)

segment_summary['revenue_share_pct'] = (
    segment_summary['total_revenue']
    / segment_summary['total_revenue'].sum()
    * 100
)

print(
    segment_summary.to_string(index=False)
)

print("\n========== VALIDATION ==========")

print(
    "Total sellers in segmentation:",
    seller_segmentation['seller_id'].nunique()
)

print(
    "Segment count:",
    seller_segmentation['seller_segment'].nunique()
)

print(
    "Missing seller segments:",
    seller_segmentation['seller_segment'].isna().sum()
)

print(
    "Duplicate seller IDs:",
    seller_segmentation['seller_id'].duplicated().sum()
)

print(
    "Revenue reconciliation:",
    seller_segmentation['product_revenue'].sum(),
    "vs",
    risk_base['product_revenue'].sum()
)

print(
    "Revenue difference:",
    seller_segmentation['product_revenue'].sum()
    - risk_base['product_revenue'].sum()
)
```
